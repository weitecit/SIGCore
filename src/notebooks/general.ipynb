```python
import geopandas as gpd

gdf = gpd.read_file(r'C:\Users\danie\Downloads\unicos_historico.geojson')
```

```python
target_columns = [s for s in gdf.columns.tolist() if 'left' in s]
target_columns.extend(['geometry', 'field'])
#remove '_left' from column names
new_columns = [s.replace('_left', '') for s in target_columns]
gdf_renamed = gdf[target_columns].rename(columns=dict(zip(target_columns, new_columns)))
gdf_renamed.fillna(0, inplace=True)
```

```python
import mongo_api as mapi

features = mapi._gdf_to_mongo_structure(gdf_renamed)
new_features = mapi._apply_layer_model(features, 'plot', 'undefined', 'undefined')
```

```python
new_features[0]['properties']
```

```python
result = mapi.db.layers.insert_many(new_features)
print(f"Inserted {len(result.inserted_ids)} documents")
```
