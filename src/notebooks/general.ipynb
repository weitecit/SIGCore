```python
import mongo_api as mapi
from datetime import datetime

gdf = mapi.get_parcelario(client='SMT')
gdf = gdf[gdf['created_datetime'] > datetime(2026, 8, 1)]
gdf_group = gdf.dissolve('parcel_id', aggfunc={'dn_surface': 'sum', 'parcel':'first'})
gdf_filtered = gdf_group[gdf_group['dn_surface']< 16000]
```

```python
gdf_filtered['parcel'].values
```

```python
gdf_filtered.to_file('filtered_parcelario.geojson', driver='GeoJSON')
```
