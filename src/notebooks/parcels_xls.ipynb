```python
import mongo_api as mapi

gdf = mapi.get_parcelario()
```

```python
dis_gdf = gdf.dissolve(by='parcel',
                       aggfunc= 
{
            "dn_surface": "sum"
        }
).reset_index()
dis_gdf['centroid'] = dis_gdf.geometry.centroid
dis_gdf['area'] = dis_gdf['dn_surface']/10000
dis_gdf['centro'] = dis_gdf['centroid'].apply(lambda x: (x.x, x.y))
```

```python
dis_gdf[['parcel', 'area', 'centro']].to_excel('Parcelario.xlsx', index=False)
```
